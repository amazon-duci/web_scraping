#### Getting some the pages in the Jumia site

```python
from time import sleep

link = 'https://www.jumia.com.ng/'
url_list = ['mlp-appliances/', 'phone-tablets', 'health-beauty/', 'home-office/']
page_part1 = '?page='
page_part2 = '#catalog-listing'

data = []

for cat in url_list:
    for num in range(5):
        url = f"{link}{cat}{page_part1}{num}{page_part2}"
        response = requests.get(url)
        html_content = response.text
        soup = bs(html_content, 'lxml')
        all_page_items = soup.find_all('div', class_ = 'itm col')

        for item in all_page_items:    
            item_name = item.find('div', class_ = 'name')
            name = item_name.text if item_name else 'NA' # returns 'NA' if it doesn't exist
            
            item_price = item.find('div', class_ = 'prc')
            price = item_price.text if item_price else 'NA'
            
            item_discount = item.find('div', class_ = 'bdg _dsct')
            discount = item_discount.text if item_discount else 'NA'
            
            item_stock = item.find('div', class_ = 'stk')
            stock = item_stock.text if item_stock else 'NA'
        
            goods = {
                'name': name,
                'price': price,
                'discount': discount,
                'stock': stock
            }
        
            data.append(goods)
        sleep(2) # after each iteration, it waits for 2 secs, (after importing sleep from time)
        
```

## Scraping JiJi site

```python
jiji_url = 'http://jiji.ng/vehicles'
jiji_response = requests.get(jiji_url)
jiji_content = jiji_response.text
jiji_soup = bs(jiji_content, 'lxml')

jiji_data = []

item_collection = jiji_soup.find_all('div', class_ = 'masonry-column')

for item in item_collection:
    jiji_items = item.find_all('div', class_ = 'masonry-item')

    for good in jiji_items:
        product_name = good.find('div', class_ = 'qa-advert-title')
        name = product_name.text.strip() if product_name else 'NA'

        product_price = good.find('div', class_ = 'qa-advert-price')
        price = product_price.text.strip() if product_price else 'NA'

        product_describe = good.find('div', 'b-list-advert-base__description-text')
        description = product_describe.text.strip() if product_describe else 'NA'

        product_img = good.find('img')
        image_link = product_img.get('src') if product_img else 'NA'

        goods = {
            'Names': name,
            'Price': price,
            'Description': description,
            'Image_Link': image_link
        }
        jiji_data.append(goods)
```

```python
jiji_data
```

```python
df = pd.DataFrame(jiji_data)
df.head()
```

```python
# Extracting the Brand Name and storing in a column
df['Brand'] = df['Names'].apply(lambda x : str(x).split(' ')[0])
df.head()
```

```python
# Formatting the price to be mathematically operational
def clean(text):
    return text.replace(' ', '').replace('₦', '').replace(',', '')


df['Selling_Price'] = df['Price'].apply(lambda x: float(clean(str(x))) + 500000)  
# We cleaned the value and added 500,000 to the price and converted to float
df.head(10)
```
